## 1. Diccionarios

### Ejercicios

**1.1** Crea un diccionario `yo` con tres claves, `nombre`, `ciudad` y `edad`, con tus datos. Enséñalo.

**1.3** Ha pasado un año. Suma 1 a tu edad y vuelve a enseñar el diccionario.

**1.4** Añade una clave `comida` con tu comida favorita.

## 2. JSON y la API

Enlaces:

- https://akabab.github.io/starwars-api/api/id/1.json
- https://pokeapi.co/api/v2/pokemon/pikachu

### Ejercicios

**2.1** Pide a la API el personaje número 4 y guárdalo en un diccionario `vader`. Enseña su nombre y su altura.

Resultado esperado: Darth Vader, 2.03

**2.2** Enseña la foto de Darth Vader.

**2.3** Elige un número del 1 al 88 y enseña el nombre, la especie y la foto de ese personaje. Escribe en el chat quién te ha salido. Si el código de la respuesta es 404, prueba con otro número.

**2.4** ¿Qué pasa si pides el personaje número 1000? Enseña solo el código de la respuesta.
