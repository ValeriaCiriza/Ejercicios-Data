# Python desde cero

Variables, condicionales, listas, bucles y funciones.

## 0. Comprobación

## 1. Variables

### Ejercicios

**1.1** Quedan 2 litros de agua y sois 4 personas. Calcula cuántos litros le tocan a cada uno.

Resultado esperado: 0.5

**1.2** Tienes 14 tablones y cada ventana necesita 3. Calcula cuántas ventanas puedes tapiar y cuántos tablones sobran.

Resultado esperado: 4 ventanas y sobran 2 tablones

## 2. Condicionales: if, elif, else

### Ejercicios

**2.1** Crea la variable `temperatura = 39.5`. Si es mayor que 38, muestra "Fiebre: te han mordido. ¡A cuarentena!". Si no, muestra "Estás sano. Puedes entrar". Prueba también con 36.5.

**2.2** La puerta del refugio solo se abre si la clave es "zanahoria" y la persona no está mordida. Crea las variables `clave` y `mordido` y muestra "Adelante" o "No pasas". Prueba las tres combinaciones: clave incorrecta, persona mordida y todo correcto.

**2.3** (opcional) Según el número de zombis, muestra el nivel de amenaza:

- 0: "Tranquilidad"
- de 1 a 5: "Precaución"
- de 6 a 20: "Peligro"
- más de 20: "HORDA: ¡corred!"

Prueba con 0, 4, 15 y 50.

## 3. Listas

### Ejercicios

**3.1** Crea una lista `mochila` con "linterna", "agua" y "cuerda". Añade "botiquín" y muestra la lista, cuántos objetos tiene y cuál es el primero.

**3.2** Si "botiquín" está en la mochila, muestra "Puedes curarte". Si no, muestra "Sin botiquín". Después quítalo con `remove` y vuelve a comprobarlo.

## 4. Bucle for

### Ejercicios

**4.1** Con `grupo = ["Lucía", "Carmen", "Andrés", "Toby"]`, muestra una línea por persona: "Lucía presente", "Carmen presente", etc.

**4.2** Cada día gastamos este agua: `litros_por_dia = [1.5, 2, 0.5, 3, 1]`. Calcula el total con un for, sin usar `sum`. Si el total es menor que 10, muestra "Hay que buscar agua".

Resultado esperado: 8.0

**4.3** (opcional) `zombis_por_dia = [3, 8, 2, 15, 6]`. Sin usar `max`, calcula cuántos zombis hubo el peor día y qué día fue.

Resultado esperado: día 4, con 15 zombis

## 5. Bucle while y números aleatorios

### Ejercicios

**5.1** Con un while, cuenta hacia atrás desde 10 hasta 1 mostrando "Faltan X días". Al terminar, muestra "¡RESCATE!".

**5.2** Empiezas con `comida = 0`. Mientras tengas menos de 10 latas, sales a explorar y encuentras `random.randint(0, 3)` latas. Cuenta cuántas salidas necesitas.

**5.3** (opcional) La clave del búnker es un número al azar del 1 al 20. Pide números con `input()` hasta que el jugador acierte, diciéndole si la clave es más alta o más baja. Al final, muestra cuántos intentos ha necesitado.

## 6. Funciones

### Ejercicios

**6.1** Escribe una función `litros_para(personas, dias)` que devuelva los litros necesarios, sabiendo que cada persona bebe 2 litros al día. Pruébala con 4 personas y 3 días, y con 10 personas y 7 días.

Resultado esperado: 24 y 140

**6.2** Convierte el ejercicio 2.3 en una función `nivel_amenaza(zombis)` que devuelva el texto con `return`. Úsala en un for con cada número de la lista `[0, 4, 15, 50]`.

**6.3** (opcional) Escribe una función `contar_seises(veces)` que tire un dado tantas veces como indique `veces` y devuelva cuántos seises han salido. Pruébala con 6, 60 y 6000.
