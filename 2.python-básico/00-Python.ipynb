### Listas

Las listas son uno de los datos más frecuentes en Python, en buena medida atribuído al uso de lo que se conocen como *listas de comprensión*, que veremos más adelante. Se inicializan mediante corchetes, y sus elementos se separan por comas. Una operación muy frecuente en las listas es el añadir elementos al final de la misma, para lo cual usamos el método *.append()*:

Para manejar los elementos de las listas, tendremos qeu saber cómo indexarlas. En el caso más elemental, los podemos obtener individualmente introduciendo al final de la misma entre corchetes el índice del elemento que deseamos recuperar *(**en Python, los índices comienzan por el valor cero**)*:

## Funciones

Es posible que esté familiarizado con el concepto matemático de una función . Una función es una relación o mapeo entre una o más entradas y un conjunto de salidas. En matemáticas, una función normalmente se representa así:

$$ z = f(x,y)$$

Aquí, fay una función $f$ que opera en las entradas , $x$ e $y$. La salida de la función es $z$. Sin embargo, las funciones de programación son mucho más generalizadas y versátiles que esta definición matemática. De hecho, la definición y el uso apropiados de funciones son tan críticos para el desarrollo adecuado de software que prácticamente todos los lenguajes de programación modernos admiten funciones integradas y definidas por el usuario.

En programación, una función es un bloque de código autónomo que encapsula una tarea específica o un grupo de tareas relacionadas. En los tutoriales anteriores de esta serie, se le presentaron algunas de las funciones integradas proporcionadas por Python. La función `id()`, por ejemplo, toma un argumento y devuelve el identificador entero único de ese objeto:

````python
s = 'string'
id(s)
````

`len()` devuelve la longitud del argumento que se le pasó:

````python
a = ['foo', 'bar', 'baz', 'qux']
len(a)
````

`any()` toma un iterable como argumento y devuelve `True` si alguno de los elementos en el iterable es verdadero y `False` de lo contrario:

````python
any(['bar' == 'baz', len('foo') == 3, 'qux' in {'foo', 'bar', 'baz'}])
````

Cada una de estas funciones integradas realiza una tarea específica. El código que realiza la tarea está definido en alguna parte, pero no es necesario saber dónde ni cómo funciona el código. Todo lo que necesita saber es la interfaz de la función:

* Qué argumentos (si los hay) se necesitan.
* Qué valores (si los hay) devuelve.

Tras ello, llama a la función y pasa los argumentos apropiados. La ejecución del programa va al cuerpo de código designado y hace su trabajo útil. Cuando finaliza la función, la ejecución vuelve a su código donde lo dejó. La función puede o no devolver datos para que su código los use, como lo hacen los ejemplos anteriores.

Cuando define su propia función de Python, funciona de la misma manera. Desde algún lugar de su código, llamará a su función de Python y la ejecución del programa se transferirá al cuerpo del código que conforma la función. Cuando finaliza la función, la ejecución vuelve a la ubicación donde fue llamada. Dependiendo de cómo haya diseñado la interfaz de la función, los datos pueden pasarse cuando se llama a la función y los valores devueltos pueden devolverse cuando finaliza.

### La importancia de las funciones en Python

Prácticamente todos los lenguajes de programación que se usan hoy en día admiten una forma de funciones definidas por el usuario, aunque no siempre se las llama funciones. En otros lenguajes, es posible que los vea como uno de los siguientes:

* Subrutinas
* Procedimientos
* Métodos
* Subprogramas

Entonces, ¿por qué molestarse en definir funciones?

* Abstracción y reutilización: La abstracción de la funcionalidad en una definición de función es un ejemplo del principio de desarrollo de software *Don't Repeat Yourself* (DRY). Podría decirse que esta es la motivación más fuerte para usar funciones.
* Modularidad: Las funciones permiten dividir procesos complejos en pasos más pequeños. En lugar de unir todo el código, se divide en funciones separadas, cada una de las cuales se enfoca en una tarea específica.
* Uso de *namespaces*: Un *namespace* es una región de un programa en la que los identificadores tienen significado. Como verá a continuación, cuando se llama a una función de Python, se crea un nuevo *namespace* para esa función, uno que es distinto de todos los demás espacios de nombres que ya existen. El resultado práctico de esto es que las variables se pueden definir y usar dentro de una función de Python incluso si tienen el mismo nombre que las variables definidas en otras funciones o en el programa principal. En estos casos, no habrá confusión ni interferencia porque se mantienen en espacios de nombres separados.

### Llamadas de función y definición

La sintaxis habitual para definir una función de Python es la siguiente:

```python
def <function_name>([<parameters>]):
    <statement(s)>
```
Veamos lo que representa cada uno:

|                 |                                                                                                            |
|-----------------|------------------------------------------------------------------------------------------------------------|
| `def `            | La palabra clave que informa a Python que se está definiendo una función                                   |
| `<function_name>` | Un identificador de Python válido que nombra la función                                                   |
|  `<parameters>`   | Una lista opcional de parámetros separados por comas que se pueden pasar a la función                      |
| `:`              | Puntuación que indica el final del encabezado de la función de Python (el nombre y la lista de parámetros) |
| `<statement(s)>`  | Un bloque de sentencias de Python válidas                                                                  |


El elemento final, `<statement(s)>`, se llama el cuerpo de la función. El cuerpo es un bloque de declaraciones que se ejecutarán cuando se llame a la función. El cuerpo de una función de Python se define mediante sangría de acuerdo con la *off-side rule*.

Comenzaremos con un sencillo ejemplo, al que iremos incrementando la complejidad:

````python
def f():
    s = '-- Inside f()'
    print(s)

print('Before calling f()')
f()
print('After calling f()')
````

Así es como funciona este código:

* La línea 1 usa la defpalabra clave para indicar que se está definiendo una función. La ejecución de la defdeclaración simplemente crea la definición de f(). Todas las siguientes líneas que están sangradas (líneas 2 a 3) pasan a formar parte del cuerpo de f()y se almacenan como su definición, pero aún no se ejecutan.

* La línea 4 es un pequeño espacio en blanco entre la definición de la función y la primera línea del programa principal. Si bien no es sintácticamente necesario, es bueno tenerlo.

* La línea 5 es la primera declaración que no tiene sangría porque no es parte de la definición de f(). Es el comienzo del programa principal. Cuando se ejecuta el programa principal, esta declaración se ejecuta primero.

* La línea 6 es una llamada a `f()`. Tenga en cuenta que siempre se requieren paréntesis vacíos tanto en una definición de función como en una llamada de función, incluso cuando no hay parámetros o argumentos. La ejecución continúa y se ejecutan las declaraciones en el cuerpo de `f()`.

* La línea 7 es la siguiente línea a ejecutar una vez que el cuerpo de `f()` ha terminado. La ejecución vuelve a esta sentencia `print()`.

En ocasiones, es posible que desee definir una función vacía que no haga nada. Esto se conoce como *stub*, que suele ser un marcador de posición temporal para una función de Python que se implementará por completo en un momento posterior. Así como un bloque en una estructura de control no puede estar vacío, tampoco el cuerpo de una función. Para definir una función de código auxiliar, utilice la declaración `pass`:

````python
def f():
    pass

f()
````

Más adelante, al estudiar la herencia de clases, veremos que existe una forma más elegante de abordar este hecho con un tipo de excepción específico, `NotImplementedError`.

### Paso de argumentos

Hasta ahora, las funciones que ha definido no han tomado ningún argumento. Eso a veces puede ser útil, y ocasionalmente escribirá tales funciones. Sin embargo, lo más frecuente es que desee **pasar datos a una función** para que su comportamiento pueda variar de una invocación a la siguiente. Veamos cómo hacer eso.

#### Argumentos posicionales

La forma más sencilla de pasar argumentos a una función de Python es con argumentos posicionales (también llamados argumentos requeridos). En la definición de la función, especifica una lista de parámetros separados por comas dentro de los paréntesis:

````python
# Definimos función
def f(qty, item, price):
    print(f'{qty} {item} cost ${price:.2f}')
# Cuando se llama a la función, especifica una lista de argumentos:
f(6, 'bananas', 1.74)
````

Los parámetros (`qty`, `item` y `price`) se comportan como variables que se definen localmente en la función. Cuando se llama a la función, los argumentos que se pasan (`6`, `'bananas'`y `1.74`) se vinculan a los parámetros en orden, como si se tratara de una asignación de variables. En algunos textos de programación, los parámetros proporcionados en la definición de la función se denominan **parámetros formales** y los argumentos en la llamada a la función se denominan **parámetros reales**.

Aunque los argumentos posicionales son la forma más sencilla de pasar datos a una función, también ofrecen la menor flexibilidad:

* Para empezar, el orden de los argumentos en la llamada debe coincidir con el orden de los parámetros en la definición. Es responsabilidad del programador que define la función documentar cuáles deberían ser los argumentos apropiados.
* Con argumentos posicionales, los argumentos en la llamada y los parámetros en la definición deben coincidir no solo en orden sino también en número.

Veamos qué sucede en el primer caso:

````python
# Argumentos en orden incorrecto
f('bananas', 1.74, 6)
````

Del mismo modo, en el segundo casose genera una excepción, `TypeError`:

````python
# Argumentos de menos o de más
f(6, 'bananas')
````

#### Argumentos de palabras clave

Cuando llamas a una función, puedes especificar argumentos en el formulario `<keyword>=<value>`. En ese caso, cada `<keyword>` debe coincidir con un parámetro en la definición de función de Python. Por ejemplo, la función definida anteriormente `f()` se puede llamar con argumentos de palabras clave de la siguiente manera:

````python
f(qty=6, item='bananas', price=1.74)
````

Hacer referencia a una palabra clave que no coincide con ninguno de los parámetros declarados genera una excepción, nuevamente `TypeError`:

````python
f(qty=6, item='bananas', cost=1.74)
````

Sin embargo, al igual que con los argumentos posicionales, la cantidad de argumentos y parámetros aún debe coincidir. Por lo tanto, los argumentos de palabras clave permiten flexibilidad en el orden en que se especifican los argumentos de función, pero el número de argumentos sigue siendo rígido. Además, una vez que haya especificado un argumento de palabra clave, no puede haber argumentos posicionales a su derecha; esto genera un `SyntaxError`:

````python
f(6, item='bananas', 1.74)
````

#### Parámetros predeterminados

Si un parámetro especificado en una definición de función de Python tiene la forma `<name>=<value>`, `<value>` se convierte en un valor predeterminado para ese parámetro. Los parámetros definidos de esta manera se denominan parámetros predeterminados u opcionales. A continuación se muestra un ejemplo de una definición de función con parámetros predeterminados:

````python
def f(qty=6, item='bananas', price=1.74):
    print(f'{qty} {item} cost ${price:.2f}')
````

Cuando se llama a esta versión de `f()`, cualquier argumento que se omite asume su valor predeterminado:

````python
f(4, 'apples', 2.24) # incluimos todos los valores
f(item='kumquats', qty=9) # dejamos 'price' por defecto
f(price=2.29) # dejamos 'qty' e 'item' por defecto
````

#### Valores de parámetros predeterminados mutables

La situación puede complicarse si especifica un valor de parámetro predeterminado que es un objeto mutable. Considere esta definición de función de Python:

````python
def f(my_list=[]):
    my_list.append('###')
    return my_list
````

`f()` toma un solo parámetro de lista, agrega la cadena `'###'` al final de la lista y devuelve el resultado:

````python
f(['foo', 'bar', 'baz'])
````

El valor predeterminado para el parámetro `my_list` es la lista vacía, por lo que si `f()` se llama sin ningún argumento, el valor devuelto es una lista con un solo elemento `'###'`:

````python
print(f"Primera llamada de la función sin especificar argumentos: {f()}")
````

Ahora, ¿qué esperaría que sucediera si `f()` se llama sin ningún parámetro una segunda y una tercera vez? Veamos:

````python
print(f"Segunda llamada de la función sin especificar argumentos: {f()}")

print(f"Tercera llamada de la función sin especificar argumentos: {f()}")
````

Es posible que haya esperado que cada llamada posterior también devuelva la lista de singleton `['###']`, al igual que la primera. En cambio, el valor de retorno sigue creciendo.

En Python, los valores de parámetros predeterminados se definen solo una vez cuando se define la función (es decir, cuando defse ejecuta la instrucción). El valor predeterminado no se vuelve a definir cada vez que se llama a la función. Por lo tanto, cada vez que llame `f()` sin un parámetro, estará actuando `.append()` en la misma lista. Podemos comprobar que, en efecto, se trata del mismo objeto con `id()`:

````python
# Función
def f(my_list=[]):
    print(id(my_list)) # añadimos muestra por consola del identificador
    my_list.append('###')
    return my_list

# Primera llamada
print(f"Primera llamada de la función sin especificar argumentos: {f()}")

print(f"Segunda llamada de la función sin especificar argumentos: {f()}")

print(f"Tercera llamada de la función sin especificar argumentos: {f()}")
````

El **identificador de objeto** que se muestra confirma que, cuando my_listse permite el valor predeterminado, el valor es el mismo objeto con cada llamada. Dado que las listas son mutables, cada `.append()` llamada posterior hace que la lista se alargue. Este es un error común y bastante bien documentado cuando se utiliza un objeto mutable como valor predeterminado de un parámetro. Potencialmente conduce a un comportamiento confuso del código y probablemente sea mejor evitarlo.

Como solución alternativa, considere usar un valor de argumento predeterminado que indique que **no se ha especificado ningún argumento**. Casi cualquier valor funcionaría, pero `None` es una opción común. Cuando el valor centinela indica que no se proporciona ningún argumento, cree una nueva lista vacía dentro de la función:

````python
# Función
def f(my_list=None):
    if my_list is None:
        my_list = []
    my_list.append('###')
    return my_list

# Primera llamada
print(f"Primera llamada de la función sin especificar argumentos: {f()}")

print(f"Segunda llamada de la función sin especificar argumentos: {f()}")

print(f"Tercera llamada de la función sin especificar argumentos: {f()}")
````

### Declaración `return`

Uno de los principales propósitos del uso de funciones es modularizar operaciones en un código. Para ello, es frecuente estar interesados en variables intermedias o finales que se encapsulan en dicha función; lo realizaremos mediante el uso de **valores de retorno de función**. 

La declaración `return` tiene dos propósitos:

* Finalizar inmediatamente la ejecución de la función.
* Retornar datos contenidos en la misma.

#### Salida de una función

Dentro de una función, una returndeclaración provoca la salida inmediata de la función de Python y la transferencia de la ejecución de nuevo a quien la llama:

````python
def f():
    print('foo')
    print('bar')
    return
````

En este ejemplo, la declaración `return` es en realidad superflua. Una función volverá al invocador que la llama cuando se cae del final (traducido de la expresión anglosajona **falls off the end**), es decir, después de que se ejecute la última declaración del cuerpo de la función. Entonces, esta función se comportaría de manera idéntica sin la declaración `return`.

Sin embargo, las declaraciones `return` no necesitan estar al final de una función. Pueden aparecer en cualquier parte del cuerpo de una función, e incluso varias veces. Considere este ejemplo:

````python
def f(x):
    if x < 0:
        return
    if x > 100:
        return
    print(x)

for i in [-3,105,67]:
    print(f"Retorno del método para el número {i}: {f(i)}")
````

Las dos primeras llamadas a `f()` no generan ningún resultado, porque se ejecuta una instrucción `return` y la función finaliza prematuramente, antes de que se alcance la instrucción `print()` en la línea 6. Este tipo de paradigma puede ser útil para **verificar errores** en una función. 

#### Devolución de datos

Además de salir de una función, la declaración `return` también se usa para devolver datos a la persona que llama. Si una declaración `return` dentro de una función de Python va seguida de una expresión, entonces, en el entorno de llamada, la llamada de función se evalúa como el valor de esa expresión:

````python
def f():
    return 'foo'

s = f()
s
````

Aquí, el valor de la expresión `f()` en la línea 5 es `'foo'`, que posteriormente se asigna a la variable `s`.

Una función puede devolver cualquier tipo de objeto. En Python, eso significa prácticamente cualquier cosa. En el entorno de llamada, la llamada de función se puede usar sintácticamente de cualquier forma que tenga sentido para el tipo de objeto que devuelve la función.

Por ejemplo, en el siguiente código, `f()` devuelve un diccionario. Entonces, en el entorno de llamada, la expresión `f()` representa un diccionario y `f()['baz']` es una referencia clave válida en ese diccionario:

````python
def f():
    return dict(foo=1, bar=2, baz=3)

print(f())
print(f()['baz'])
````

Si se especifican varias expresiones separadas por comas en una declaración `return`, se empaquetan y se devuelven como una **tupla**:

````python
def f():
    return 'foo', 'bar', 'baz', 'qux'

print(f"Tipo de dato de salida de la función: {type(f())}")
print(f"Salida de la función: {f()}")

a, b, c, d = f()
print(f'Detalle del guardado de variables de salida: a = {a}, b = {b}, c = {c}, d = {d}')
````

Cuando no se proporciona ningún valor de retorno, o se finaliza una función con la declaración `pass`, una función de Python devuelve el valor especial `None`:

````python
def f():
    return

print(f())
````

### Argumentos de longitud variable

En algunos casos, cuando está definiendo una función, es posible que no sepa de antemano cuántos argumentos querrá que tome. Suponga, por ejemplo, que desea escribir una función de Python que calcule el promedio de varios valores. Un posible intento inicial resultaría en algo como la siguiente función:

````python
# Función
def avg(a, b, c):
    return (a + b + c) / 3

# Ejemplo de uso
avg(1,2,3)
````

Sin embargo, como ya ha visto, cuando se utilizan argumentos posicionales, la cantidad de argumentos que se pasan debe coincidir con la cantidad de parámetros declarados. Claramente, entonces, esta implementación de `avg()` resultará en error para cualquier número de valores que no sean tres:

````python
avg(1, 2, 3, 4)
````

Podríamos intentar definir `avg()` con parámetros opcionales:

```python
def avg(a, b=0, c=0, d=0, e=0):
    ...
```

Esto permite especificar un número variable de argumentos. Las siguientes llamadas son al menos sintácticamente correctas. Pero este enfoque todavía adolece de un par de problemas:

* Para empezar, solo permite hasta cinco argumentos, no un número arbitrario.
* Además, no hay forma de distinguir entre los argumentos que se especificaron y los que se permitieron por defecto.

La función no tiene forma de saber cuántos argumentos se pasaron realmente, por lo que no sabe por qué dividir:

```python
def avg(a, b=0, c=0, d=0, e=0):
    return (a + b + c + d + e) / # ¿Dividido por...?
```

Podríamos optar por poner un parámetro lista como opción, pero ya vimos que no es buena práctica poner objetos mutables como parámetros de las mismas. En este caso, la mejor opción es que Python proporciona una manera de pasar a una función un número variable de argumentos con el empaquetado y desempaquetado de tuplas de argumentos usando el *operador asterisco ( ).

#### Empaquetamiento de tuplas de argumentos

Cuando el nombre de un parámetro en una definición de función de Python está precedido por un asterisco (*), indica el empaquetamiento de tuplas de argumentos. Todos los argumentos correspondientes en la llamada a la función se empaquetan en una tupla a la que la función puede hacer referencia mediante el nombre de parámetro dado. 

````python
def f(*args):
    print(args)
    print(type(args), len(args))
    for x in args:
            print(x)

f(1, 2, 3)
````

En la definición de `f()`, la especificación del parámetro `*args` indica empaquetamiento de tuplas. En cada llamada a `f()`, los argumentos se empaquetan en una tupla a la que la función puede hacer referencia por el nombre args. Se puede usar cualquier nombre, pero `args` se elige tan comúnmente que es prácticamente un estándar. Pôdemos ahora reescribir nuestra función de media de la siguiente forma:

````python
def avg(*args):
    return sum(args) / len(args)

avg(1, 2, 3, 4, 5)
````

Aún así, dependiendo de cómo se use este código, aún puede haber trabajo por hacer. Tal como está escrito, `avg()` producirá una excepción `TypeError` si algún argumento no es numérico; estudiaremos esta cuestión en el próximo apartado.

#### Desempaquetado de tuplas de argumentos

Una operación análoga está disponible en el otro lado de la ecuación en una llamada de función de Python. Cuando un argumento en una llamada de función está precedido por un asterisco (*), indica que el argumento es una tupla que se debe desempaquetar y pasar a la función como valores separados:

````python
def f(x, y, z):
    print(f'x = {x}')
    print(f'y = {y}')
    print(f'z = {z}')

t = ('foo', 'bar', 'baz')
f(*t)
````

En este ejemplo, `*t` en la llamada a la función indica que `t` es una tupla que se debe desempaquetar. Los valores descomprimidos `'foo'`, `'bar'`y `'baz'` se asignan a los parámetros `x`, `y`, `z`, respectivamente.

Aunque este tipo de desempaquetado se llama desempaquetado de tuplas, no solo funciona con tuplas. El operador asterisco (*) se puede aplicar a cualquier iterable en una llamada de función de Python. Por ejemplo, también se puede descomprimir una lista o un conjunto:

````python
a = ['foo', 'bar', 'baz']
print(f'Uso de desempaquetado con listas:')
f(*a)

s = {1, 2, 3}
print(f'Uso de desempaquetado con tuplas:')
f(*s)
````

Incluso puede usar el empaquetado y desempaquetado de tuplas al mismo tiempo:

````python
def f(*args):
    print(type(args), args)

a = ['foo', 'bar', 'baz', 'qux']
f(*a)
````

#### Empaquetado del diccionario de argumentos

Python tiene un operador similar, el asterisco doble (`**`), que se puede usar con los parámetros y argumentos de las funciones de Python para especificar el empaquetado y desempaquetado de diccionarios . Preceder un parámetro en una definición de función de Python con un asterisco doble (`**`) indica que los argumentos correspondientes, que se espera que sean pares `key=value`, deben empaquetarse en un diccionario :

````python
def f(**kwargs):
    print(kwargs)
    print(type(kwargs))
    for key, val in kwargs.items():
            print(key, '->', val)

f(foo=1, bar=2, baz=3)
````

En este caso, los argumentos `foo=1`, `bar=2` y `baz=3` se empaquetan en un diccionario al que la función puede hacer referencia por el nombre `kwargs`. Nuevamente, se puede usar cualquier nombre, pero el peculiar `kwargs` (que es la abreviatura de palabra clave `args`) es casi estándar. No tiene que adherirse a él, pero si lo hace, cualquier persona familiarizada con las convenciones de codificación de Python sabrá de inmediato lo que quiere decir.

#### Desempaquetado del diccionario de argumentos

El desempaquetado del diccionario de argumentos es análogo al desempaquetado de tuplas de argumentos. Cuando el asterisco doble (`**`) precede a un argumento en una llamada de función de Python, especifica que el argumento es un diccionario que se debe desempaquetar, y los elementos resultantes se pasan a la función como argumentos de palabras clave:

````python
def f(a, b, c):
    print(F'a = {a}')
    print(F'b = {b}')
    print(F'c = {c}')

d = {'a': 'foo', 'b': 25, 'c': 'qux'}
f(**d)
````

Los elementos del diccionario dse desempaquetan y se pasan `f()` como argumentos de palabras clave. Entonces, `f(**d)` es equivalente a `f(a='foo', b=25, c='qux')`:

#### Poniendolo todo junto

Para combinar estas ideas, veremos `*args` como una lista de argumentos posicionales de longitud variable y `**kwargs` como una lista de argumentos de palabras clave de longitud variable. Los tres (parámetros posicionales estándar `*args`, y `**kwargs`) se pueden usar en una definición de función de Python. Si es así, entonces deben especificarse en ese orden:

````python
def f(a, b, *args, **kwargs):
    print(F'a = {a}')
    print(F'b = {b}')
    print(F'args = {args}')
    print(F'kwargs = {kwargs}')

f(1, 2, 'foo', 'bar', 'baz', 'qux', x=100, y=200, z=300)
````

Esto proporciona tanta flexibilidad como la que podría necesitar en una interfaz de funciones.

### Cadenas de documentación (`docstrings`)

Cuando la primera declaración en el cuerpo de una función de Python es un literal de cadena, se conoce como la **cadena de documentación** de la función. Una cadena de documentación se utiliza para proporcionar documentación para una función. Puede contener el propósito de la función, qué argumentos toma, información sobre los valores devueltos o cualquier otra información que crea que sería útil.

El siguiente es un ejemplo de una definición de función con una cadena de documentación:

````python
def avg(*args):
    """Returns the average of a list of numeric values."""
    return sum(args) / len(args)
````

Técnicamente, las cadenas de documentos pueden usar cualquiera de los mecanismos de comillas de Python, pero la convención recomendada es usar comillas triples usando caracteres de comillas dobles (`"""`), como se muestra arriba. Si la cadena de documentación cabe en una línea, las comillas de cierre deben estar en la misma línea que las comillas de apertura.

Las cadenas de documentos de varias líneas se utilizan para una documentación más extensa. Una cadena de documentos de varias líneas debe constar de una línea de resumen, seguida de una línea en blanco, seguida de una descripción más detallada. Las comillas de cierre deben estar en una línea por sí mismas:

```python
def foo(bar=0, baz=1):
    """Perform a foo transformation.

    Keyword arguments:
    bar -- magnitude along the bar axis (default=0)
    baz -- magnitude along the baz axis (default=1)
    """
    <function_body>
```

Las convenciones semánticas y de formato de docstring se detallan en [este recurso](https://peps.python.org/pep-0257/).

Cuando se define una cadena de documentación, el intérprete de Python la asigna a un atributo especial de la función llamada `__doc__`. Este atributo es uno de un conjunto de identificadores especializados en Python que a veces se denominan **atributos** mágicos o **métodos mágicos** porque proporcionan una funcionalidad de lenguaje especial.

````python
print(avg.__doc__)
````

Se considera una buena práctica de codificación especificar una cadena de documentación para cada función de Python que defina.

## Gestión de excepciones

Un programa de Python termina tan pronto como encuentra un error. En Python, un error puede ser un error de sintaxis o una excepción.

Los errores de sintaxis ocurren cuando el analizador detecta una declaración incorrecta. Observa el siguiente ejemplo:

````python
print( 0 / 0 ))
````

La flecha indica dónde se encontró el analizador con el **error de sintaxis**. En este ejemplo, había un paréntesis de más. Elimínelo y ejecute su código nuevamente:

````python
print( 0 / 0)
````

Esta vez, se encontró con un **error de excepción**. Este tipo de error ocurre siempre que el código Python sintácticamente correcto da como resultado un error. La última línea del mensaje indicaba qué tipo de error de excepción se encontró.

En lugar de mostrar el mensaje `exception error`, Python detalla qué tipo de error de excepción se encontró. En este caso, fue un `ZeroDivisionError`. Python viene con varias excepciones integradas, así como con la posibilidad de crear excepciones autodefinidas.

### Generación de una excepción

Podemos usar `raise` para lanzar una excepción si ocurre una condición. La declaración se puede complementar con una excepción personalizada. Si desea arrojar un error cuando se produce una determinada condición usando `raise`, puede hacerlo de esta manera:

````python
x = 10
if x > 5:
    raise Exception('x no debe ser mayor que 5. Su valor es {}'.format(x))
````

El programa se detiene y muestra nuestra excepción en la pantalla, ofreciendo pistas sobre lo que salió mal.

### Excepción `AssertionError`

En lugar de esperar a que un programa se bloquee a mitad de camino, también puede comenzar haciendo una verificación en Python. Tenemos `assert` que una determinada condición se cumple. Si esta condición resulta ser `True`, el programa puede continuar. Si la condición resulta ser `False`, puede hacer que el programa lance una `AssertionErrorexcepción`.

Eche un vistazo al siguiente ejemplo, donde se afirma que el código se ejecutará en un sistema Linux: Si ejecuta este código en una máquina Linux, la verificación se supera. Si tuviera que ejecutar este código en una máquina con Windows, el resultado de la afirmación sería `False` y el resultado sería el siguiente:

````python
import sys
assert ('linux' in sys.platform), "This code runs on Linux only."
````

### Bloques `try` y `except`

El binomio `try`-`except` en Python se usa para capturar y manejar excepciones, controlando el código de igual manera qeu lo haría un `if`-`else`. Python ejecuta el código siguiendo la declaración `try` como una parte "normal" del programa. El código que sigue a la declaración `except` es la respuesta del programa a cualquier excepción en la cláusula `try` anterior.

Cuando el código sintácticamente correcto se encuentra con un error, Python generará un error de excepción. Este error de excepción bloqueará el programa si no se controla. La cláusula `except` determina cómo responde su programa a las excepciones.

````python
# Método auxiliar
def linux_interaction():
    assert ('linux' in sys.platform), "Function can only run on Linux systems."
    print('Doing something.')

# Código
try:
    linux_interaction()
except:
    pass
````

Aquí vemos cómo se gestionan las excepciones con estas cláusulas:
* Por una parte, el método `linux_interaction()` solo se puede ejecutar en un sistema Linux. El `assert` en esta función lanzará una `AssertionErrorexcepción` si lo llama en un sistema operativo que no sea Linux.
* Posteriormente, la forma en que manejó el error aquí es entregando un archivo pass. Si tuviera que ejecutar este código en una máquina con Windows, obtendría un resultado vacío, y en particular el programa no se bloqueará. Pero sería bueno ver si se produce algún tipo de excepción cada vez que ejecuta su código.

````python
try:
    linux_interaction()
except:
    print('Linux function was not executed')
````

Cuando ocurre una excepción en un programa que ejecuta esta función, el programa continuará y le informará sobre el hecho de que la llamada a la función no tuvo éxito.

Lo que no pudo ver fue el tipo de error que se produjo como resultado de la llamada a la función. Para ver exactamente qué salió mal, debe detectar el error que arrojó la función.

El siguiente código es un ejemplo en el que captura `AssertionError` y envía ese mensaje a la pantalla:

````python
try:
    linux_interaction()
except AssertionError as error:
    print(error)
    print('The linux_interaction() function was not executed')
````

El primer mensaje es el `AssertionError`, que le informa que la función solo se puede ejecutar en una máquina Linux. El segundo mensaje le dice qué función no se ejecutó.

Más allá de las verificaciones de código, existen muchas otras casuísticas en las que podemos proporcionar un alto grado de detalle en la gestión de errores. Un ejemplo es el siguiente:

````python
try:
    with open('file.log') as file:
        read_data = file.read()
except:
    print('Could not open file.log')
````

Este es un mensaje informativo, y nuestro programa seguirá ejecutándose. En la [documentción de Python](https://docs.python.org/3/library/exceptions.html), puede ver que hay muchas excepciones integradas que puede usar aquí. Una excepción descrita en esa página es la siguiente:

````python
try:
    with open('file.log') as file:
        read_data = file.read()
except FileNotFoundError as fnf_error:
    print(fnf_error)
````

Puede tener más de una llamada de función en su cláusula `try` y anticipar la captura de varias excepciones. Una cosa a tener en cuenta aquí es que el código en la cláusula `try` se detendrá tan pronto como se encuentre una excepción. En cualquier caso, el uso de la cláusula `except` no es recomendado si no se especifica algún tipo detallado de error, pues puede ser contraproducente y omitir detalles de errores desconocidos.

Combinemos ahora los dos casos que hemos visto:

````python
try:
    linux_interaction()
    with open('file.log') as file:
        read_data = file.read()
except FileNotFoundError as fnf_error:
    print(fnf_error)
except AssertionError as error:
    print(error)
    print('Linux linux_interaction() function was not executed')
````

En el caso de que este codigo no se ejecutase en una máquina Linux, retornaría:

```shell
Function can only run on Linux systems.
Linux linux_interaction() function was not executed
```

Dentro de la cláusula `try`, se encontró con una excepción inmediatamente y no llegó a la parte en la que intenta abrir el fichero. Sin embargo, al ejecutar este código en una máquina Linux, tendríamos:

```shell
[Errno 2] No such file or directory: 'file.log'
```

### Cláusula `else`

En Python, usando la declaración `else`, puede indicarle a un programa que ejecute un determinado bloque de código solo en ausencia de excepciones. Veámoslo sobre un ejemplo, pero esta vez ajustándonos a nuestro entorno ed ejecución (`win32`):

````python
# Método auxiliar
def win32_interaction():
    assert ('win32' in sys.platform), "Function can only run on Linux systems."
    print('Doing something.')

try:
    win32_interaction()
except AssertionError as error:
    print(error)
else:
    print('Executing the else clause.')
````

Debido a que el programa no se encontró con ninguna excepción, se ejecutó la cláusula `else`.

También puede ejecutar código `try`-`except` dentro de la cláusula `else` y detectar posibles excepciones:

````python
try:
    win32_interaction()
except AssertionError as error:
    print(error)
else:
    try:
        with open('file.log') as file:
            read_data = file.read()
    except FileNotFoundError as fnf_error:
        print(fnf_error)
````

Desde la salida, puede ver que la función `win32_interaction()` se ejecutó. Debido a que no se encontraron excepciones, se intentó abrir `file.log`. Ese archivo no existía y, en lugar de abrirlo, detectó la excepción `FileNotFoundError`.

### Cláusula `finally`

Para concluir esta sección, y a modo de broche a nuestra cadena de control de excepciones, exponemos la cláusula que nos permite, bajo cualquiera de los escenarios anteriores, asegurarnos de que se ejecuta una pieza de código. Típicamente se usa para limpieza de memoria.

````python
try:
    win32_interaction()
except AssertionError as error:
    print(error)
else:
    try:
        with open('file.log') as file:
            read_data = file.read()
    except FileNotFoundError as fnf_error:
        print(fnf_error)
finally:
    print('Cleaning up, irrespective of any exceptions.')
````

### Resumen

Después de ver la diferencia entre los errores de sintaxis y las excepciones, aprendimos varias formas de generar, capturar y manejar excepciones en Python. Concretamente:

* `raise` le permite lanzar una excepción en cualquier momento.
* `assert` le permite verificar si se cumple una determinada condición y lanzar una excepción si no es así.
* En la cláusula `try`, todas las declaraciones se ejecutan hasta que se encuentra una excepción.
* `except` se usa para capturar y manejar las excepciones que se encuentran en la cláusula `try`.
* `else` le permite codificar secciones que deben ejecutarse solo cuando no se encuentran excepciones en la cláusula de prueba.
* `finally` le permite ejecutar secciones de código que siempre deberían ejecutarse, con o sin excepciones encontradas previamente.

## Programación orientada a objetos (OOP) en Python

La programación orientada a objetos es un paradigma de programación que proporciona un medio para estructurar programas de modo que las propiedades y los comportamientos se agrupan en objetos individuales. Por ejemplo, un objeto podría representar a una persona con propiedades como nombre, edad y dirección y comportamientos como caminar, hablar, respirar y correr. O podría representar un correo electrónico con propiedades como una lista de destinatarios, asunto y cuerpo y comportamientos como agregar archivos adjuntos y enviar.

Dicho de otra manera, la programación orientada a objetos es un enfoque para modelar cosas concretas del mundo real, como automóviles, así como relaciones entre cosas, como empresas y empleados, estudiantes y profesores, etc. OOP modela entidades del mundo real como objetos de software que tienen algunos datos asociados y pueden realizar ciertas funciones.

Otro paradigma de programación común es la programación procedimental, que estructura un programa como una receta en el sentido de que proporciona un conjunto de pasos, en forma de funciones y bloques de código, que fluyen secuencialmente para completar una tarea.

La conclusión clave es que los objetos están en el centro de la programación orientada a objetos en Python, no solo representando los datos, como en la programación procedimental, sino también en la estructura general del programa.

### Clases

Las estructuras de datos **primitivas**, como números, cadenas y listas, están diseñadas para representar información simple, como el costo de una manzana, el nombre de un poema o sus colores favoritos, respectivamente. ¿Qué pasa si quieres representar algo más complejo?

Por ejemplo, supongamos que desea realizar un seguimiento de los empleados de una organización. Debe almacenar información básica sobre cada empleado, como su nombre, edad, puesto y el año en que comenzó a trabajar.

Una forma de hacer esto es representar a cada empleado como una lista:

````python
kirk = ["James Kirk", 34, "Captain", 2265]
spock = ["Spock", 35, "Science Officer", 2254]
mccoy = ["Leonard McCoy", "Chief Medical Officer", 2266]
````

Hay una serie de problemas con este enfoque:

* Primero, puede hacer que los archivos de código más grandes sean más difíciles de administrar. Si hace referencia a `kirk[0]` varias líneas de donde se declara la lista `kirk`, ¿recordará que el elemento con índice `0` es el nombre del empleado?

* En segundo lugar, puede introducir errores si no todos los empleados tienen el mismo número de elementos en la lista. En la lista anterior `mccoy`, falta la edad, por lo que `mccoy[1]` regresará `"Chief Medical Officer"` en lugar de la edad del Dr. McCoy.

Una excelente manera de hacer que este tipo de código sea más manejable y fácil de mantener es usar **clases**. Las clases se utilizan para crear estructuras de datos definidas por el usuario. Las clases definen funciones llamadas **métodos**, que identifican los comportamientos y acciones que un objeto creado a partir de la clase puede realizar con sus datos. En definitiva, una clase es un modelo de cómo debe definirse algo.

Mientras que la clase es el modelo, una instancia es un objeto que se crea a partir de una clase y contiene datos reales. Dicho de otra manera, una clase es como un formulario o cuestionario. Una instancia es como un formulario que se ha llenado con información.

Todas las definiciones de clase comienzan con la palabra clave `class`, seguida del nombre de la clase y dos puntos. Cualquier código que esté sangrado debajo de la definición de la clase se considera parte del cuerpo de la clase.

````python
class HumanBeing:
    pass
````

El cuerpo de la clase `HumanBeing` consiste en una sola declaración: la palabra clave `pass`. Esta a menudo se usa como un marcador de posición que indica dónde irá el código eventualmente. Le permite ejecutar este código sin que Python arroje un error.

La clase `HumanBeing` no es muy interesante en este momento, así que mejorémosla un poco definiendo algunas propiedades que todos los objetos `HumanBeing` deberían tener. Hay una serie de propiedades entre las que podemos elegir, incluido el nombre, la edad o el género.

Las propiedades que deben tener todos los objetos se definen en un método llamado `.__init__()`. Cada vez que se crea un nuevo objeto `HumanBeing`, establece el estado `.__init__()` inicial del objeto asignando los valores de las propiedades del objeto. Es decir, inicializa cada nueva instancia de la clase `.__init__()`.

Puede dar `.__init__()` cualquier número de parámetros, pero el primer parámetro siempre será una variable llamada `self`. Cuando se crea una nueva instancia de clase, la instancia se pasa automáticamente al parámetro `self` para que se puedan definir nuevos **atributos** en el objeto `.__init__()`.

Actualicemos la clase `HumanBeing` con un método `.__init__()` que crea los parámetros comentados anteriormente:

````python
class HumanBeing:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
````

Los atributos creados en `.__init__()` se denominan **atributos de instancia**. El valor de un atributo de instancia es específico de una instancia particular de la clase. Todos los objetos `HumanBeing` tienen un nombre, edad y género, pero los valores de los atributos variarán según la instancia concreta `HumanBeing`.

Por otro lado, los **atributos de clase** son atributos que tienen el mismo valor para todas las instancias de clase. Puede definir un atributo de clase asignando un valor a un nombre de variable fuera de `.__init__()`.

````python
class HumanBeing:
    
    # Atributos de clase
    isHumanBeing = True

    def __init__(self, name, age, gender):
        # Atributos de instancia
        self.name = name
        self.age = age
        self.gender = gender
````

### Instancias

La creación de un nuevo objeto a partir de una clase se denomina **instanciación de un objeto**. Puede instanciar un nuevo objeto `HumanBeing` escribiendo el nombre de la clase, seguido de paréntesis de apertura y cierre:

````python
# Ejemplo de juguete
class HumanBeing:
    pass

# Instanciación
HumanBeing()
````

Esta cadena de letras y números de aspecto divertido es una dirección de memoria que indica dónde `HumanBeing` se almacena el objeto en la memoria de su computadora. Tenga en cuenta que la dirección que ve en su pantalla será diferente.

Ahora crea una instancia de un segundo objeto `HumanBeing`:

````python
HumanBeing()
````

La nueva instancia `HumanBeing` se encuentra en una dirección de memoria diferente. Esto se debe a que es una instancia completamente nueva y es completamente única desde el primer objeto `HumanBeing` que se creó.

### Atributos de clase e instancia

Retornando al ejemplo completo de clase `HumanBeing`, podemos comprobar que si no especificamos los atributos de instancia obtendremos un `TypeError`:

````python
# Clase
class HumanBeing:
    
    # Atributos de clase
    isHumanBeing = True

    def __init__(self, name, age, gender):
        # Atributos de instancia
        self.name = name
        self.age = age
        self.gender = gender

# Instanciación
HumanBeing()
````

Para pasar argumentos a los parámetros, colocaremos los valores entre paréntesis después del nombre de la clase:

````python
person = HumanBeing("John", 38, "Male")
````

El método `.__init__()` de la clase `HumanBeing` tiene cuatro parámetros, entonces, ¿por qué solo se le pasan tres argumentos?

Cuando crea una instancia de un objeto `HumanBeing`, Python crea una nueva instancia y la pasa al primer parámetro de `.__init__()`. Básicamente, esto elimina el parámetro `self`, por lo que solo debe preocuparse por los parámetros `name`, `age` y `gender`.

Después de crear la instancia `HumanBeing`, puede acceder a sus atributos de instancia mediante la **notación de puntos**:



````python
# Atributos de instancia
print(f"Atributos de instancia:\n\t# Nombre: {person.name}\n\t# Edad: {person.age}\n\t# Género: {person.gender}")
# Atributos de clase
print(f"Atributos de clase:\n\t# isHumanBeing: {person.isHumanBeing}")
````

Una de las mayores ventajas de usar clases para organizar datos es que se garantiza que las instancias tendrán los atributos que espera. Todas las instancias `HumanBeing` tienen atributos `name`, `age`, `gender` y `isHumanBeing`, por lo que puede usar esos atributos con confianza sabiendo que siempre devolverán un valor.

Aunque se garantiza que los atributos existen, sus valores se pueden cambiar dinámicamente:

````python
person.age = 45
print(f"Nuevo valor del atributo de instancia edad: {person.age}")
````

### Métodos de instancia

Los métodos de instancia son funciones que se definen dentro de una clase y solo se pueden llamar desde una instancia de esa clase. Al igual que `.__init__()`, el primer parámetro de un método de instancia siempre es `self`.

````python
# Clase
class HumanBeing:
    
    # Atributos de clase
    isHumanBeing = True

    def __init__(self, name, age, gender):
        # Atributos de instancia
        self.name = name
        self.age = age
        self.gender = gender
    
    # Método de instancia sin argumentos propios
    def describe(self):
        print(f"{self.name} es un {self.gender} de {self.age} años.")
    
    # Método de instancia con argumentos propios
    def speak(self, text):
        print(f"{self.name} dice: {text}")

# Ejemplo
person = HumanBeing("John", 38, "varón")
person.describe()
person.speak('Estoy estudiando Inteligencia Artificial.')
````

Al escribir sus propias clases, es una buena idea tener un método que devuelva una cadena que contenga información útil sobre una instancia de la clase. Sin embargo, `.description()` no es la forma más *elegante* de hacer esto. En su lugar, se emplea un **método dunder** como `.__init__()`, conocido como `.__str__()`. Así, con un simple `print` sobre el objeto que instancia la clase, obtendremos la misma información:

````python
# Clase
class HumanBeing:
    
    # Atributos de clase
    isHumanBeing = True

    def __init__(self, name, age, gender):
        # Atributos de instancia
        self.name = name
        self.age = age
        self.gender = gender
    
    # Método dunder
    def __str__(self):
        return f"{self.name} es un {self.gender} de {self.age} años."
    
    # Método de instancia con argumentos propios
    def speak(self, text):
        print(f"{self.name} dice: {text}")

# Ejemplo
person = HumanBeing("John", 38, "varón")
print(person)
person.speak('Estoy estudiando Inteligencia Artificial.')
````

### Herencia de clases

La herencia es el proceso por el cual una clase adquiere los atributos y métodos de otra. Las clases recién formadas se denominan **clases secundarias**, y las clases de las que se derivan las clases secundarias se denominan **clases principales**.

Las clases secundarias pueden anular, sobrescribir o ampliar los atributos y métodos de las clases principales. En otras palabras, las clases secundarias heredan todos los atributos y métodos de los padres, pero también pueden especificar atributos y métodos que son únicos para ellos mismos. Por ejemplo, imaginemos que como parte de `HumanBeing` queremos especificar a los *data scientist*, indicando en un método de instancia adicional cuál es su tecnología favorita, y sobrescribiendo el método `speak` especificando un mensaje por defecto:

````python
# Clase principal
class HumanBeing:
    
    # Atributos de clase
    isHumanBeing = True

    def __init__(self, name, age, gender):
        # Atributos de instancia
        self.name = name
        self.age = age
        self.gender = gender
    
    # Método dunder
    def __str__(self):
        return f"{self.name} es un {self.gender} de {self.age} años."
    
    # Método de instancia con argumentos propios
    def speak(self, text):
        print(f"{self.name} dice: {text}")

# Clase secundaria
class DataScientist(HumanBeing):
    # Actualizamos método speak
    def speak(self, text:str='¡Hola! Soy data scientist'):
        return f"{self.name} dice: {text}"
    # Añadimos un método de tecnología favorita
    def favTech(self, tech:str):
        return f"La tecnología favorita de {self.name} es {tech}"

# Ejemplo
person = DataScientist("John", 38, "varón")
print(person.speak())
print(person.favTech("python"))
````

La actualización que hemos hecho al método `speak` es bastante *liviana*, y de hecho en circunstancias normales no requeriría sobreescribir un método entero. Para ser más eficientes, se utiliza `super()` para obtener el método de instancia de la clase principal, y usarlo de una forma concreta:

````python
# Clase secundaria
class DataScientist(HumanBeing):
    # Actualizamos método speak de manera eficiente
    def speak(self, text:str='¡Hola! Soy data scientist'):
        return super().speak(text)
    # Añadimos un método de tecnología favorita
    def favTech(self, tech:str):
        return f"La tecnología favorita de {self.name} es {tech}"
````

De hecho, si queremos que una clase principal imponga como condición necesaria que se genere una clase secundaria específica en cada caso, podemos usar la excepción `NotImplementedError`:

````python
# Clase principal
class HumanBeing:
    
    # Atributos de clase
    isHumanBeing = True

    def __init__(self, name, age, gender):
        # Atributos de instancia
        self.name = name
        self.age = age
        self.gender = gender
    
    # Método dunder
    def __str__(self):
        return f"{self.name} es un {self.gender} de {self.age} años."
    
    # Método de instancia con argumentos propios
    def speak(self, text):
        raise NotImplementedError("Debe heredar a una clase e implementar este método de forma nativa")

# Clase secundaria
class DataScientist(HumanBeing):
    # Actualizamos método speak
    def speak(self, text:str='¡Hola! Soy data scientist'):
        return f"{self.name} dice: {text}"
    # Añadimos un método de tecnología favorita
    def favTech(self, tech:str):
        return f"La tecnología favorita de {self.name} es {tech}"
````
